```python
import pandas as pd

# 1. Main debt data
df_data = pd.read_csv("IDS_ALLCountries_Data.csv", encoding="latin1")
print("Main data:", df_data.shape)
display(df_data)
df_data.describe()
```

```python
df_data.info()
```

```python
#drop columns from 2025 t0 2032:
df_data.drop(df_data.loc[:, '2025':'2032'].columns, axis=1, inplace=True)
print(df_data.columns.tolist()) #check if the columns are droped


```

```python
print("Shape:", df_data.shape)
print("\nColumns:", df_data.columns.tolist())
print("\nData types:\n", df_data.dtypes)
print("\nMissing values per column:\n", df_data.isnull().sum())
print("\nDuplicate rows:", df_data.duplicated().sum())
```

```python
 #Drop useless columns
df_data = df_data.drop(columns=['Counterpart-Area Name', 'Counterpart-Area Code'])
#Drop duplicate rows
df_data = df_data.drop_duplicates()
print("Duplicates:", df_data.duplicated().sum())
print(df_data.isnull().sum())
```

```python
#Drop rows with missing identifiers (broken rows)
df_data = df_data.dropna(subset=['Country Name', 'Country Code'])
print("Shape after cleanup:", df_data.shape)
```

```python
year_cols = [str(y) for y in range(2000, 2025)]
id_cols = ['Country Name', 'Country Code', 'Series Name', 'Series Code']

long_df = df_data.melt(
    id_vars=id_cols,
    value_vars=year_cols,
    var_name='Year',
    value_name='Value'
)

# Drop rows where there's no value for that year (this is where the ~13k/year nulls get handled)

long_df['Year'] = long_df['Year'].astype(int)

# Sort so years are in order within each group before filling
long_df = long_df.sort_values(['Country Name', 'Series Code', 'Year'])

# Forward-fill Value within each Country + Series group only
long_df['Value'] = long_df.groupby(['Country Name', 'Series Code'])['Value'].ffill()

#backward fill value after forward fill to drop null
long_df['Value'] = long_df.groupby(['Country Name', 'Series Code'])['Value'].bfill()

#Handle series with zero historical data across all years
long_df['Value'] = long_df['Value'].fillna(0)

display("Long format shape:", long_df.shape)
display(long_df.head(5))
print("Nulls:\n", long_df.isnull().sum())
```

```python
long_df.to_csv("IDS_ALLCountries_Data_cleaned.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")

# load cleaned data
df = pd.read_csv("IDS_ALLCountries_Data_cleaned.csv")
df["Country Code"] = df["Country Code"].str.strip()
df["Series Code"] = df["Series Code"].str.strip()

# Filter out World Bank regional/income aggregates (not real countries)
AGGREGATE_CODES = ['EAP', 'ECA', 'IDX', 'IDA', 'LAC', 'LDC', 'LMY',
                    'LIC', 'LMC', 'MNA', 'MIC', 'SAS', 'SSA', 'UMC',
                    'WLD', 'HIC', 'OED', 'PST', 'HPC', 'EAS',
                    'EAR', 'ECS', 'EMU', 'LTE', 'MEA', 'NAC', 'OSS',
                    'PRE', 'PSS', 'TEA', 'TEC', 'TLA', 'TMN', 'TSA', 'TSS']
df = df[~df["Country Code"].isin(AGGREGATE_CODES)]
print("Rows after filtering out aggregates:", df.shape)

# Key headline indicator: Total External Debt Stocks
TOTAL_DEBT = "External debt stocks, total (DOD, current US$)"
debt_total = df[df["Series Name"] == TOTAL_DEBT]
print("Rows for Total External Debt indicator:", debt_total.shape)
```

```python
latest_year = 2023
debt_latest = debt_total[debt_total["Year"] == latest_year]
country_debt = debt_latest.groupby("Country Name")["Value"].sum().sort_values(ascending=False)
```

1.country-wise debt distribution

```python

import pandas as pd
import plotly.express as px

# Convert country_debt Series to DataFrame with 'Country Name' and 'Value' columns
# This variable 'country_debt' is available from previously executed cells.
country_debt_df = country_debt.reset_index(name='Value')

# Top 10 countries by debt
top15 = country_debt_df.head(15)

fig_top15 = px.bar(top15, x="Country Name", y="Value", title=f"Top 15 Countries by Total External Debt ({latest_year})",
              labels={"Value": "Debt (current US$)"}, color_discrete_sequence=["steelblue"],
              hover_data={"Value": ":,.0f"})
fig_top15.update_layout(xaxis_tickangle=-45)
fig_top15.show()

```

2.top 10 vs bottom 10 countries

```python
import pandas as pd
import plotly.express as px
# 2. Top / bottom countries
# Convert country_debt Series to DataFrame with 'Country Name' and 'Value' columns
country_debt_df = country_debt.reset_index(name='Value')

top10 = country_debt_df.head(10)
bottom10 = country_debt_df[country_debt_df["Value"] > 0].tail(10)

fig2a = px.bar(top10, x="Country Name", y="Value", title=f"Top 10 Countries by Debt ({latest_year})",
              labels={"Value": "Debt (current US$)"}, color_discrete_sequence=["steelblue"])
fig2a.update_layout(xaxis_tickangle=-45)
fig2a.show()

fig2b = px.bar(bottom10, x="Country Name", y="Value", title=f"Bottom 10 Countries by Debt ({latest_year})",
               labels={"Value": "Debt (current US$)"}, color_discrete_sequence=["salmon"])
fig2b.update_layout(xaxis_tickangle=-45)
fig2b.show()
```

3.different debt indicators and their impact

```python
import plotly.express as px

# 3. Different debt indicators and their impact

print("3. DEBT INDICATORS AND THEIR IMPACT")

# 3a. Magnitude of actual DEBT-related indicators only (exclude GNI/Imports/Exports
#     which are economic context, not debt itself)
debt_keywords = ["debt", "DOD", "TDS", "arrears"]
debt_indicators_only = df[df["Series Name"].str.contains("debt", case=False, na=False)]
indicator_totals = (
    debt_indicators_only[debt_indicators_only["Year"] == latest_year]
    .groupby("Series Name")["Value"]
    .sum()
    .abs()
    .sort_values(ascending=False)
)
print("\nTop 10 debt-specific indicators by magnitude:")
print(indicator_totals.head(10))

# Convert Series to DataFrame for Plotly Express
df_top_indicators = indicator_totals.head(10).reset_index()

fig_3a = px.bar(
    df_top_indicators,
    x="Value",
    y="Series Name",
    orientation="h",
    color_discrete_sequence=["darkorange"],
    title=f"Top 10 Debt Indicators by Total Magnitude ({latest_year})",
    labels={
        "Value": "Total Value (absolute, current US$)",
        "Series Name": "Debt Indicator",
    },
)

# Invert y-axis to match matplotlib's top-down ranking
fig_3a.update_layout(
    yaxis={"autorange": "reversed"},
    template="plotly_white",
    margin=dict(l=150, r=20, t=50, b=50),
)
fig_3a.show()
# Save to interactive HTML file if needed:
# fig_3a.write_html("eda_2_top_indicators.html")


# 3b. IMPACT: Debt-to-GNI ratio - shows how burdensome debt actually is per country
gni = df[(df["Series Name"] == "GNI (current US$)") & (df["Year"] == latest_year)]
gni = gni[["Country Name", "Value"]].rename(columns={"Value": "GNI"})

debt_vs_gni = debt_latest[["Country Name", "Value"]].rename(
    columns={"Value": "Debt"}
)
impact_df = debt_vs_gni.merge(gni, on="Country Name")
impact_df["Debt_to_GNI_pct"] = (impact_df["Debt"] / impact_df["GNI"]) * 100
impact_df = impact_df.sort_values("Debt_to_GNI_pct", ascending=False)

print(
    "\nTop 10 countries by Debt-to-GNI ratio (most debt-burdened relative to economy size):"
)
print(impact_df[["Country Name", "Debt_to_GNI_pct"]].head(10).to_string(index=False))

fig_3b = px.bar(
    impact_df.head(10),
    x="Country Name",
    y="Debt_to_GNI_pct",
    color_discrete_sequence=["firebrick"],
    title=f"Top 10 Countries by Debt-to-GNI Ratio ({latest_year}) - Debt Burden Impact",
    labels={
        "Debt_to_GNI_pct": "Debt as % of GNI",
        "Country Name": "Country",
    },
    text_auto=".2f",  # Displays percentage labels on top of bars
)

fig_3b.update_layout(
    xaxis={"tickangle": -45}, template="plotly_white", margin=dict(t=50, b=100)
)
fig_3b.show()
# Save to interactive HTML file if needed:
# fig_3b.write_html("eda_2b_debt_to_gni.html")
```

4.Patterns , trends,relationship

```python

# 4. Patterns, trends, AND relationships

#Global total external debt trend

yearly_trend = debt_total[debt_total["Year"] <= 2023].groupby("Year")["Value"].sum().reset_index()

fig4 = px.line(
    yearly_trend, x="Year", y="Value", markers=True,
    title="Global Total External Debt Trend (2000-2023)",
    labels={"Value": "Debt (current US$)"}
)
fig4.show()

top5_countries = country_debt_df.head(5)["Country Name"].tolist()
top5_trend = debt_total[(debt_total["Country Name"].isin(top5_countries)) & (debt_total["Year"] <= 2023)]

#Top 5 Most Indebted Countries (2000-2023)"
fig4a = px.line(
    top5_trend, x="Year", y="Value", color="Country Name", markers=True,
    title="External Debt Trend: Top 5 Most Indebted Countries (2000-2023)",
    labels={"Value": "Debt (current US$)"}
)
fig4a.show()

# 4b. RELATIONSHIPS: correlation between Debt, GNI, and Reserves
reserves = df[(df["Series Name"] == "Total reserves (includes gold, current US$)") & (df["Year"] == latest_year)]
reserves = reserves[["Country Name", "Value"]].rename(columns={"Value": "Reserves"})
relationship_df = impact_df.merge(reserves, on="Country Name", how="left")

corr_debt_gni = relationship_df["Debt"].corr(relationship_df["GNI"])
arr_debt_reserves = relationship_df["Debt"].corr(relationship_df["Reserves"])
print(f"Correlation between Debt and GNI: {corr_debt_gni:.3f}")
print(f"Correlation between Debt and Reserves: {arr_debt_reserves:.3f}")

fig4b = px.scatter(
    relationship_df, x="GNI", y="Debt", hover_name="Country Name",
    trendline="ols",
    title=f"Relationship: External Debt vs GNI ({latest_year}) — corr = {corr_debt_gni:.2f}",
    labels={"GNI": "GNI (current US$)", "Debt": "External Debt (current US$)"}
)
fig4b.show()
```

5.statistical summaries and comparision

```python
import plotly.express as px

# 5. Statistical summaries and comparisons

print("5. STATISTICAL SUMMARY (Total External Debt, 2023)")

print(debt_latest["Value"].describe())

print("\nStandard deviation:", debt_latest["Value"].std())
print("Median:", debt_latest["Value"].median())
print("Skew:", debt_latest["Value"].skew())

# 5a. Boxplot to visualize spread/outliers
fig_5a = px.box(
    debt_latest,
    x="Value",
    points="outliers",
    title=f"Distribution of Total External Debt Across Countries ({latest_year})",
    labels={"Value": "Debt (current US$)"},
    color_discrete_sequence=["steelblue"],
)
fig_5a.update_layout(template="plotly_white", margin=dict(t=50, b=50))
fig_5a.show()
#fig_5a.write_html("eda_5_boxplot.html")


# 5b. COMPARISONS: across Region and Income Group
countries_meta = pd.read_csv("IDS_CountryMetaData_cleaned.csv")
countries_meta = countries_meta[
    ~countries_meta["Country Code"].str.strip().isin(AGGREGATE_CODES)
]

debt_with_meta = debt_latest.merge(countries_meta, on="Country Name", how="left")

print("\nAverage debt by Region:")
region_avg = (
    debt_with_meta.groupby("Region")["Value"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
print(region_avg.to_string(index=False))

# Bar chart by Region
fig_5b = px.bar(
    region_avg,
    x="Region",
    y="Value",
    title=f"Average External Debt by Region ({latest_year})",
    labels={"Value": "Average Debt (current US$)", "Region": "Region"},
    color_discrete_sequence=["teal"],
    text_auto=".2s",  # Shows formatted numbers (e.g., $1.2B) on top of bars
)
fig_5b.update_layout(
    xaxis={"tickangle": -45}, template="plotly_white", margin=dict(t=50, b=100)
)
fig_5b.show()
#fig_5b.write_html("eda_5b_region_comparison.html")


print("\nAverage debt by Income Group:")
income_avg = (
    debt_with_meta.groupby("Income Group")["Value"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
print(income_avg.to_string(index=False))

# Bar chart by Income Group
fig_5c = px.bar(
    income_avg,
    x="Income Group",
    y="Value",
    title=f"Average External Debt by Income Group ({latest_year})",
    labels={"Value": "Average Debt (current US$)", "Income Group": "Income Group"},
    color_discrete_sequence=["purple"],
    text_auto=".2s",
)
fig_5c.update_layout(
    xaxis={"tickangle": -45}, template="plotly_white", margin=dict(t=50, b=100)
)
fig_5c.show()
#fig_5c.write_html("eda_5c_income_comparison.html")

```
